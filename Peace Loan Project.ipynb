## Default Risk Analysis

```python
# Question 6. Do people who default have a lower credit score
print("Average Credit Score for each group:")
print(loan.groupby('Defaulted')['Credit_Score'].mean())

# Simple chart to show it visually
plt.figure(figsize=(6, 4))
sns.barplot(data=loan, x='Defaulted', y='Credit_Score', ci=None, palette='Set2')
plt.title('Credit_Score: Paid (No) vs Defaulted (Yes)')
plt.xlabel('Defaulted Status')
plt.ylabel('Credit_Score')
plt.show()
```

```python
# Question 7: Does a borrower's employment status affect their likelihood of defaulting?
plt.figure(figsize=(8, 4))
# Counts defaults across different employment types
sns.countplot(data=loan, x='Employment_Status', hue='Defaulted', palette='viridis')
plt.title('Defaults Across Employment Statuses')
plt.xlabel('Employment_Status')
plt.ylabel('Number of Customers')
plt.legend(title='Defaulted?')
plt.xticks(rotation=15)
plt.show()
```

# Customer Segmentation

```python
# Question 9: How many of our current borrowers fall into High, Medium, and Low risk tiers?
# Create the risk tiers using a simple rule function
def assign_risk(row):
    if row['Credit_Score'] < 600 or row['Defaulted'] == 'Yes':
        return 'High Risk'
    elif 600 <= row['Credit_Score'] <= 700:
        return 'Medium Risk'
    else:
        return 'Low Risk'

# Apply the rules to create a new column
loan['Risk_Segment'] = loan.apply(assign_risk, axis=1)

# Print out the exact customer count in each tier
print("Total Customers per Risk Tier:")
print(loan['Risk_Segment'].value_counts())

# Draw a clean bar chart of our risk distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=loan, x='Risk_Segment', order=['Low Risk', 'Medium Risk', 'High Risk'], palette='coolwarm')
plt.title('CrediTrust Portfolio Risk Breakdown')
plt.xlabel('Risk Segment Category')
plt.ylabel('Number of Borrowers')
plt.show()
```

```python
Question 9: Which loan purposes are our "High Risk" segments mostly applying for?
plt.figure(figsize=(9, 5))
# Shows the distribution of loan reasons specifically inside our risk segments
sns.countplot(data=loan, x='Loan_Purpose', hue='Risk_Segment', palette='magma')
plt.title('Loan Purposes Demanded by Each Risk Tier')
plt.xlabel('Reason for Loan')
plt.ylabel('Number of Applications')
plt.xticks(rotation=30)
plt.legend(title='Risk Tier')
plt.show()
```
